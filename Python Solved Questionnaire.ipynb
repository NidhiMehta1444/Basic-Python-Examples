## Write a program to find whether given number is armstrong number

```python
n=int(input("Enter the number:"))
s=0
num=n
while(n>0):
    r=n%10
    s=s+(r**3)
    n=n/10
if(s==num):
    print("Armstrong")
else:
    print("Not Armstrong")
```

# Write a Python program to find duplicate elements. (VERY IMPORTANT)

```python
numbers=[1,2,3,2,1,4]
seen=set()
duplicates=set()

for n in numbers:
    if n in seen:
        duplicates.add(n)
    else:
        seen.add(n)
print(duplicates)
```
